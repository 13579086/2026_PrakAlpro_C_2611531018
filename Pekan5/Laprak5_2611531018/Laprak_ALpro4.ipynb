## 1.1 Latar Belakang

Dalam dunia pemrograman, sebuah program jarang sekali berjalan secara linier dari awal hingga akhir tanpa adanya pertimbangan kondisi. ## 2.1 Dasar Teori

Percabangan adalah struktur kontrol yang memungkinkan instruksi dieksekusi hanya jika suatu kondisi bernilai True (benar). Jika kondisi bernilai False (salah), instruksi tersebut dilewati atau digantikan oleh instruksi alternatif.

Jenis-Jenis Percabangan pada Python

1. If Tunggal: Digunakan jika hanya ada satu aksi yang dieksekusi saat kondisi bernilai True.

2. If-Else: Digunakan untuk menangani dua kemungkinan aksi (True dan False).

3. If-Elif-Else: Digunakan ketika terdapat lebih dari dua kondisi yang saling mengeksklusifkan (hanya satu blok kondisi yang dieksekusi).

4. Multi-IF Terpisah: Rangkaian if independen di mana setiap kondisi dievaluasi secara terpisah tanpa terikat oleh kondisi sebelumnya, memungkinkan beberapa blok kode tereksekusi secara bersamaan/akumulatif.

5. Match-Case: Fitur struktur pencocokan variabel terhadap pola atau nilai tertentu yang membuat penulisan kode lebih rapi dibandinkan rantai if-elif panjang.

## 2.2 Kode Program

### 2.2.1 Program "if1_2611531018.py"

```python
# Buat file dengan nama if1_nim.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()  

ipk_1018 = float(input("Masukkan IPK Anda: ")) 
 
if ipk_1018 > 2.75:
    print("Anda Lulus Sangat Memuaskan dengan IPK " + str(ipk_1018))
 
print("Program Selesai")
```

**Penjelasan :**

Program ini mengevaluasi nilai IPK pengguna menggunakan bentuk percabangan paling sederhana, yaitu if tunggal.

**Alur Logika:**

1. Program menerima input IPK dari pengguna yang dikonversi ke tipe data desimal (float).

2. Kondisi if ipk_1018 > 2.75 mengecek apakah nilai yang dimasukkan melebihi 2,75.

3. Jika kondisi bernilai True, pesan kelulusan disajikan di layar.

4. Jika kondisi bernilai False, program langsung melewati instruksi di dalam blok if.

5. Pernyataan print("Program Selesai") akan selalu dieksekusi di akhir karena berada di luar blok if.

### 2.2.2 Program "if2_2611531018.py"

```python
# Buat file dengan nama if2_nim.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()  

ipk_1018 = float(input("Masukkan IPK Anda: ")) 
 
if ipk_1018 > 2.75:
    print("Anda Lulus Sangat Memuaskan dengan IPK " + str(ipk_1018))
else:
    print("Anda tidak lulus")
print("Program Selesai")
```

**Penjelasan :**

Program ini mengikutsertakan jalan keluar alternatif menggunakan struktur if-else.

**Alur Logika:**

1. Program mengevaluasi nilai IPK inputan.

2. Jika IPK lebih dari 2.75, program mencetak keterangan lulus.

3. Jika IPK tidak memenuhi kriteria tersebut (<= 2.75), alur program otomatis melompat ke blok else dan mencetak "Anda tidak lulus".

4. Struktur ini menjamin salah satu dari dua blok pesan pasti dijalankan sebelum program berakhir.

### 2.2.3 Program "if_elif_else1_2611531018.py"

```python
# Buat file dengan nama if_elif_else1_nim.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()

umur_1018 = int(input("Input umur anda: "))
sim_1018 = input("Apakah Anda Sudah Punya Sim C: ")[0]

if umur_1018 >= 17 and sim_1018 == 'y':
    print("Anda Sudah dewasa dan boleh bawa motor")
elif umur_1018 >= 17 and sim_1018 != 'y':
    print("Anda Sudah dewasa tetapi tidak boleh bawa motor")
elif umur_1018 < 17 and sim_1018 == 'y':
    print("Anda Belum Cukup Umur punya SIM")
else:
    print("Anda Belum Cukup Umur dan tidak boleh bawa motor")
print("Program Selesai")
```

**Penjelasan :**

Program ini menguji beberapa kombinasi variabel umur dan kepemilikan SIM menggunakan rantai if-elif-else dan operator logika and.

**Alur Logika:**

1. Menerima input umur (int) dan karakter pertama kepemilikan SIM ([0]).

2. Kondisi 1 (if): Diuji apakah umur >= 17 dan sim == 'y'. Jika keduanya benar, dicetak status dewasa dan diizinkan membawa motor.

3. Kondisi 2 (elif): Jika kondisi pertama gagal, diuji apakah umur >= 17 tetapi sim != 'y'. Menandakan umur cukup namun belum ber-SIM.

4. Kondisi 3 (elif): Diuji jika umur < 17 tetapi secara data terisi sim == 'y' (kondisi identitas/SIM belum sah).

5. Blok else: Menangani seluruh kasus tersisa, yaitu pengguna di bawah umur dan belum memiliki SIM.

6. Struktur if-elif-else bersifat mutually exclusive: setelah satu kondisi terpenuhi, kondisi berikutnya tidak akan diuji lagi.

### 2.2.4 Program "nilai_2611531018.py"

```python
# Buat file dengan nama nilai_nim.py
# Buat program untuk nested if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()
# program konversi nilai angka menjadi huruf

nilai_1018 = int(input("Inputkan nilai angka= "))

if nilai_1018 >= 81:
    print("A")
elif nilai_1018 >= 70:
    print("B")
elif nilai_1018 >= 60:
    print("C")
elif nilai_1018 >= 50:
    print("D")
else:
    print("E")
```

**Penjelasan :**

Program ini mengonversi nilai angka ke huruf mutu (grade) menggunakan pengujian tingkat hirarkis bertingkat.

**Alur Logika:**

1. Input Data: Program menerima masukan nilai angka dari pengguna menggunakan fungsi input(), yang kemudian diubah menjadi tipe data bilangan bulat (int).   

2. Evaluasi Hirarkis: Pengecekan kondisi dilakukan bertingkat menggunakan struktur if-elif-else dari batas nilai tertinggi ke nilai terendah:   

if nilai_1018 >= 81: Jika nilai 81 atau lebih, program mencetak huruf mutu A.   

elif nilai_1018 >= 70: Jika nilai di antara 70 hingga 80, program mencetak huruf mutu B.   

elif nilai_1018 >= 60: Jika nilai di antara 60 hingga 69, program mencetak huruf mutu C.   

elif nilai_1018 >= 50: Jika nilai di antara 50 hingga 59, program mencetak huruf mutu D.   

else: Jika nilai kurang dari 50, program secara otomatis mencetak huruf mutu E.   

3. Sifat Mutually Exclusive: Karena menggunakan rantai if-elif-else, ketika salah satu syarat kondisi terpenuhi (misalnya masukan bernilai 75), program langsung mengeksekusi cetak huruf B lalu keluar dari blok percabangan tanpa menguji kondisi-kondisi di bawahnya lagi.

### 2.2.5 Program "multi_if1_2611531018.py"

```python
# Buat file dengan nama multi_if1_nim.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()

umur_1018 = int(input("Input umur anda: "))
sim_1018 = input("Apakah Anda Sudah Punya Sim C (y/t): ")[0]

if umur_1018 >= 17 and sim_1018 == 'y':
    print("Anda Sudah dewasa dan boleh bawa motor")

if umur_1018 >= 17 and sim_1018 != 'y':
    print("Anda Sudah dewasa tetapi tidak boleh bawa motor")

if umur_1018 < 17 and sim_1018 == 'y':
    print("Anda Belum Cukup Umur punya SIM")

if umur_1018 < 17 and sim_1018 != 'y':
    print("Anda Belum Cukup Umur bawa motor")    
```

**Penjelasan :**

Program ini melakukan fungsi yang serupa dengan Latihan 3, namun menggunakan Multi-IF terpisah (setiap pernyataan if berdiri sendiri).

**Alur Logika:**

1. Setiap baris if dievaluasi satu per satu secara independen.

2. Walaupun salah satu syarat telah terpenuhi (misalnya kondisi pertama True), Python akan tetap mengevaluasi kondisi if ke-2, ke-3, dan ke-4.

3. Karena syarat kondisinya saling menolak secara logika, hanya satu blok yang mencetak hasil, tetapi secara performa execution time, program tetap memeriksa seluruh blok.


### 2.2.6 Program "multi_if2_2611531018.py"

```python
# Buat file dengan nama multi_if2_nim.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: total_belanja_1234
# Program ini menggunakan fungsi input()
# Program Menghitung Diskon Belanja

# Input dari user
total_belanja_1018 = float(input("Masukkan total belanja (Rp): "))

# Input status member (mengecek apakah user mengetik 'y' atau 'ya')
input_member_1018 = input("Apakah Anda member? (y/t): ").strip().lower()
is_member_1018 = input_member_1018 in ["y", "ya"]

# Input status kode promo (mengecek apakah user mengetik 'y' atau 'ya')
input_promo_1018 = input("Apakah kode promo valid? (y/t): ").strip().lower()
kode_promo_valid_1018 = input_promo_1018 in ["y", "ya"]

total_diskon_persen_1018 = 0

# Multi-IF terpisah: Setiap kondisi diperiksa secara independen
# Diskon bisa ditumpuk (akumulasi) jika memenuhi beberapa syarat sekaligus

if total_belanja_1018 > 1000000:
    total_diskon_persen_1018 += 10  # Diskon belanja besar

if is_member_1018:
    total_diskon_persen_1018 += 5   # Diskon member

if kode_promo_valid_1018:
    total_diskon_persen_1018 += 15  # Diskon voucher

# Menghitung nominal diskon dan total bayar
nominal_diskon_1018 = total_belanja_1018 * (total_diskon_persen_1018 / 100)
total_bayar_1018 = total_belanja_1018 - nominal_diskon_1018

# Output hasil
print("\n--- Rincian Pembayaran ---")
print(f"Total Diskon  : {total_diskon_persen_1018}% (Rp {nominal_diskon_1018:,.0f})")
print(f"Total Bayar   : Rp {total_bayar_1018:,.0f}")

print(f"Total diskon yang Anda dapatkan: {total_diskon_persen_1018}%")
# Output: Total diskon yang Anda dapatkan: 30% jika belanja > 1 juta, member, dan kode promo valid
```

**Penjelasan :**

Program ini memanfaatkan keunggulan Multi-IF terpisah untuk membuat perhitungan diskon penumpukan (akumulatif).

**Alur Logika:**

1. Mengolah masukan total belanja, status keanggotaan, dan kode promo.

2. Variabel total_diskon_persen_1018 diinisialisasi awal dengan nilai 0.

3. if 1: Jika belanja $> 1.000.000$, persentase diskon bertambah 10%.

4. if 2: Jika user adalah member, persentase diskon bertambah 5%.

5. if 3: Jika memiliki kode promo valid, persentase diskon bertambah 15%.

6. Karena setiap if bekerja sendiri, pengguna bisa memperoleh kombinasi diskon (0%, 5%, 10%, 15%, 20%, 25%, hingga maksimal 30%).

7. Nominal diskon dan total bayar dihitung secara rinci dan ditampilkan menggunakan format angka desimal.

### 2.2.7 Program "match_case1_2611531018.py"

```python
# Buat file dengan nama match_case1.py
# Buat program untuk match case
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()
# program konversi angka menjadi nama bulan

bulan_1018 = int(input("Masukkan angka bulan (1 - 12): "))

match bulan_1018:
    case 1:
        print("Januari")
    case 2:
        print("Februari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Angka tidak valid")
```

**Penjelasan :**

Program ini menggunakan fitur match-case (pencocokan pola/pattern matching) pada Python untuk memetakan input angka menjadi nama bulan.   

**Alur Logika:**

1. Program menerima masukan angka bulan (1 - 12).   

2. Variabel bulan_1018 dicocokkan nilainya dengan tiap case.   

3. Jika nilai variabel cocok dengan angka tertentu (misal 1), instruksi di dalam case 1: yang mencetak "Januari" akan dijalankan.   

4. Blok case _: bertindak sebagai nilai bawaan (wildcard/default) yang dieksekusi jika masukan pengguna berada di luar rentang 1-12 (misal input 15 atau -1).
