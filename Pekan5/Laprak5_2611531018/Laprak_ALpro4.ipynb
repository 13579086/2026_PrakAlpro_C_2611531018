## 2.1 Dasar Teori

### 2.1.1 Perulangan for dan Fungsi range()

Perulangan for dalam Python berfungsi untuk melakukan iterasi atau penelusuran terhadap suatu urutan data (sequence). Untuk mengeksekusi blok kode dalam jumlah tertentu, for umumnya dikombinasikan dengan fungsi range(). Fungsi range() menghasilkan deret angka secara berurutan, dengan format range(start, stop, step). Struktur ini sangat efisien untuk perulangan yang batas atau jumlah iterasinya sudah diketahui secara pasti.

### 2.1.2 Perulangan Bersarang (Nested Loop)

Perulangan bersarang (nested loop) adalah kondisi di mana terdapat sebuah perulangan di dalam struktur perulangan lainnya. Perulangan yang berada di luar (outer loop) biasanya bertugas mengatur perpindahan baris, sedangkan perulangan di dalamnya (inner loop) bertugas mengatur pencetakan tiap kolom pada baris tersebut. Inner loop akan dieksekusi hingga selesai untuk setiap satu kali putaran outer loop. Konsep ini sangat krusial dalam mencetak pola dua dimensi seperti persegi, segitiga, maupun jam pasir.

### 2.1.3 Manipulasi Fungsi print()

Secara bawaan, fungsi print() di Python akan selalu mengakhiri keluarannya dengan baris baru (newline). Dalam pembuatan pola visual menggunakan nested loop, karakter sering kali harus dicetak berjejer dalam satu baris. Untuk mencegah perpindahan baris otomatis, digunakan modifikasi parameter end. Contohnya, perintah print("*", end="") akan mencetak karakter bintang tanpa baris baru, sehingga hasil cetakan selanjutnya akan berada tepat di sebelah kanan karakter tersebut.

## 2.2 Kode Program

### 2.2.1 Program "perulangan_for1_2611531018.py"

```python
# Buat file dengan nama perulangan_for1_1018.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1018
# Program ini menggunakan fungsi input()

ulang_1018 = int(input("Masukkan jumlah perulangan: "))

for i_1018 in range(ulang_1018):
    print(f"Perulangan ke-{i_1018}")
```

**Penjelasan :**

Program ini merupakan bentuk implementasi paling dasar dari struktur kontrol perulangan for dalam bahasa Python. Program berfungsi untuk mencetak teks secara berulang ke layar sebanyak nilai batas yang diinputkan oleh pengguna. Selain itu, program ini mendemonstrasikan sifat bawaan (default) dari fungsi range() di Python, yaitu memulai indeks perhitungan dari angka 0, bukan 1.

**Alur Logika:**

3. Komputer mulai mengeksekusi blok perulangan menggunakan sintaks for i_1018 in range(ulang_1018).

### 2.2.3 Program "perulangan_for3_2611531018.py"

```python
# Buat file dengan nama perulangan_for3_1018.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1018
# Program ini menggunakan fungsi input()

ulang_1018 = int(input("Masukkan jumlah perulangan: "))

jumlah_1018 = 0
for i_1018 in range(1, ulang_1018 + 1):
    print(i_1018, end=" ")
    jumlah_1018 = jumlah_1018 + i_1018

    if i_1018 < ulang_1018:
        print("+ ", end="")
    else:
        print(" = ", jumlah_1018, end="")
        
print()
print("Jumlah = ", jumlah_1018)
```

**Penjelasan :**

Program mengombinasikan struktur perulangan for dengan penggunaan variabel akumulator untuk menghitung total penjumlahan deret angka dari 1 hingga nilai batas yang diinputkan pengguna (1 + 2 + 3 + ... + n). Program ini mendemonstrasikan bagaimana perulangan tidak hanya digunakan untuk mencetak teks, tetapi juga untuk melakukan pemrosesan data aritmetika secara kumulatif pada setiap iterasi.

**Alur Logika:**

1. Program mengambil masukan berupa nilai batas dari pengguna melalui input(), mengubah tipe datanya menjadi integer, lalu menyimpannya dalam variabel batas_1018.

3. Komputer mengeksekusi perulangan for i_1018 in range(1, batas_1018 + 1) yang berjalan sekuensial dari angka 1 hingga nilai batas_1018.

4. Pada setiap putaran iterasi, nilai i_1018 ditambahkan ke dalam nilai jumlah_1018 saat ini melalui operasi akumulasi (jumlah_1018 = jumlah_1018 + i_1018).

### 2.2.4 Program "nested_for1_2611531018.py"

```python
# Buat file dengan nama nested_for1_1018.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1018
# Program ini menggunakan fungsi input()

batas_1018 = int(input("Masukkan nilai batas: "))

for line_1018 in range(1, batas_1018 + 1):
    for j_1018 in range(1, (-1 * line_1018 + batas_1018) + 1):
        print(".", end="")
    print(line_1018)
```

**Penjelasan :**

Program ini merupakan latihan dasar pengenalan struktur nested loop (perulangan bersarang) dalam bahasa Python. Program ini bertujuan untuk memberikan pemahaman dasar mengenai bagaimana dua buah perulangan for bekerja secara bertingkat untuk memproses data dua dimensi (baris dan kolom). Perulangan bagian luar (outer loop) bertindak sebagai pengatur jumlah baris secara vertikal, sedangkan perulangan bagian dalam (inner loop) bertindak sebagai pengatur pencetakan karakter atau kolom secara horizontal pada setiap barisnya.

**Alur Logika:**

### 2.2.6 Program "nested_for3_2611531018.py"

```python
# Buat file dengan nama nested_for4_1018.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1018
# Program ini menggunakan fungsi input()

tinggi_1018 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_1018 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_1018 = tinggi_1018
    c_1018 = a_1018
    lebar_1018 = (2 * tinggi_1018) - 2

    for i_1018 in range(1, tinggi_1018 + 1):
        b_1018 = c_1018 + 1

        for j_1018 in range(1, lebar_1018 + 1):

            # Baris atas dan bawah
            if i_1018 == 1 or i_1018 == tinggi_1018:
                if j_1018 == 1 or j_1018 == lebar_1018:
                    print("*", end="")
                else:
                    print("-", end="")

            # Baris isi
            else:
                if j_1018 == 1 or j_1018 == lebar_1018:
                    print("|", end="")
                else:
                    if j_1018 == c_1018:
                        print("<", end="")
                    elif j_1018 == b_1018:
                        print(">", end="")
                    elif j_1018 == (lebar_1018 - c_1018):
                        print("<", end="")
                    elif j_1018 == (lebar_1018 - c_1018 + 1):
                        print(">", end="")
                    elif j_1018 > b_1018 and j_1018 < (lebar_1018 - c_1018):
                        print(".", end="")
                    else:
                        print(" ", end="")

        print()

        # Logika asli Java
        a_1018 -= 2

        if a_1018 <= 0:
            c_1018 = (-a_1018) + 2
        else:
            c_1018 = a_1018
```

**Penjelasan :**

Program ini merupakan implementasi nested loop (perulangan bersarang) untuk membentuk pola visual segitiga piramida menggunakan karakter bintang (*). Program ini memanfaatkan satu *outer loop* untuk mengontrol jumlah baris (tinggi segitiga) dan dua inner loop yang dieksekusi secara berurutan pada setiap barisnya: inner loop pertama mencetak spasi pendorong di sebelah kiri, sedangkan inner loop kedua mencetak karakter bintang.

**Alur Logika:**

1. Program meminta input berupa tinggi segitiga dari pengguna melalui perintah input("Masukkan tinggi segitiga: "), mengubah tipe datanya menjadi integer menggunakan int(), lalu menyimpannya dalam variabel tinggi_1018.

2. Komputer mengeksekusi perulangan luar (outer loop) for i_1018 in range(1, tinggi_1018 + 1) yang bertugas mengatur urutan baris dari baris ke-1 hingga baris ke-tinggi_1018.

3. Pada setiap baris aktif, komputer terlebih dahulu menjalankan inner loop pertama for j_1018 in range(tinggi_1018 - i_1018) untuk mencetak spasi kosong (print(" ", end="")). Jumlah spasi akan berkurang secara bertahap seiring bertambahnya baris i_1018, sehingga mendorong letak bintang tepat ke posisi tengah.

4. Setelah pencetakan spasi pada baris tersebut selesai, komputer menjalankan inner loop kedua for k_1018 in range(i_1018) untuk mencetak karakter bintang yang diiringi spasi (print("*", end=" ")) sebanyak nilai baris i_1018 saat itu.

5. Setelah kedua inner loop selesai dieksekusi untuk satu baris, perintah print() tanpa argumen dipanggil di dalam outer loop untuk memindahkan kursor ke baris berikutnya (newline).

6. Alur ini terus diulang hingga seluruh baris dari outer loop selesai dijalankan, menghasilkan tampilan piramida simetris.

### 2.2.7 Program "jumlah_genap_2611531018.py"

```python
# Buat file dengan nama jumlah_genap_1018.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_1018 = int(input("Masukkan nilai batas: "))

jumlah_1018 = 0
for i_1018 in range(1, ulang_1018 + 1):
    if i_1018 % 2 == 0:
        print(i_1018, end=" ")
        jumlah_1018 = jumlah_1018 + i_1018

        if i_1018 < ulang_1018:
            print("+ ", end="")
        else:
            print("= ", jumlah_1018, end="")

print()
print("Jumlah = ", jumlah_1018)
```

**Penjelasan :**

Program ini merupakan pengembangan dari program penjumlahan deret angka. Program ini bertujuan untuk menyaring dan menjumlahkan khusus bilangan genap dari rentang angka 1 hingga nilai batas yang dimasukkan oleh pengguna. Program mendemonstrasikan penggabungan antara struktur kontrol perulangan for dengan kondisi seleksi if (operator modulus) serta variabel akumulator.

**Alur Logika:**

1. Program mengambil masukan berupa nilai batas dari pengguna melalui input(), mengonversi tipe datanya menjadi integer, lalu menyimpannya dalam variabel batas_1018.

2. Program menginisialisasi variabel penampung jumlah_1018 = 0 sebelum perulangan dimulai untuk menyimpan hasil akumulasi penjumlahan bilangan genap secara bertahap.

3. Komputer mengeksekusi perulangan for i_1018 in range(1, batas_1018 + 1) yang berjalan dari angka 1 hingga batas_1018.

4. Di dalam perulangan, dilakukan pemeriksaan kondisi if i_1018 % 2 == 0 menggunakan operator modulus (%) untuk mengecek apakah angka pada iterasi saat ini bernilai genap (sisa hasil bagi dengan 2 adalah 0).

5. Jika angka bernilai genap, nilai i_1018 ditambahkan ke dalam variabel akumulator (jumlah_1018 = jumlah_1018 + i_1018).

6. Program mencetak angka genap tersebut ke layar secara horizontal. Terdapat logika percabangan tambahan untuk mengatur pencetakan simbol + jika angka genap tersebut bukan angka genap terakhir, atau simbol = jika angka genap tersebut merupakan angka genap terakhir dalam batas.

### 2.2.8 Program "latihan5_2611531018.py"

```python
# Program untuk menampilkan pola segitiga piramida
# Nama variabel ditambah 4 digit NIM terakhir: 1018

tinggi_1018 = int(input("Masukkan tinggi segitiga: "))

for i_1018 in range(1, tinggi_1018 + 1):
    # Cetak spasi di sebelah kiri agar pola sejajar di tengah
    for j_1018 in range(tinggi_1018 - i_1018):
        print(" ", end="")
    
    # Cetak bintang beserta spasi
    for k_1018 in range(i_1018):
        print("*", end=" ")
    
    # Pindah baris
    print()
```

**Penjelasan :**

Program ini dirancang untuk mencetak pola segitiga piramida simetris menggunakan karakter bintang (*) berdasarkan masukan tinggi dari pengguna. Program ini menerapkan konsep perulangan bersarang (*nested loop*) dengan struktur satu outer loop dan dua inner loop:

* **Outer loop (i_1018):** Bertugas mengontrol urutan dan pergerakan baris secara vertikal.
* **Inner loop pertama (j_1018):** Bertugas mencetak karakter spasi pendorong di sebelah kiri agar posisi bintang sejajar di tengah.
* **Inner loop kedua (k_1018):** Bertugas mencetak karakter bintang beserta satu spasi pemisah secara horizontal.

**Alur Logika:**

1. Program menginstruksikan pengguna untuk memasukkan tinggi segitiga melalui perintah input("Masukkan tinggi segitiga: "). Masukan tersebut dikonversi menjadi tipe data integer menggunakan int() dan disimpan ke dalam variabel tinggi_1018.

2. Komputer mengeksekusi perulangan luar (outer loop) for i_1018 in range(1, tinggi_1018 + 1) untuk mengontrol baris dari baris ke-1 hingga baris ke-tinggi_1018.

Jumlah spasi yang dicetak berkurang secara bertahap seiring bertambahnya nilai baris i_1018, sehingga mendorong letak bintang ke tengah.

4. Setelah inner loop pertama selesai, komputer melanjutkan ke inner loop kedua for k_1018 in range(i_1018) untuk mencetak karakter bintang beserta spasi (print("*", end=" ")). Jumlah bintang yang dicetak pada baris tersebut sejajar dengan nilai baris i_1018 saat itu.

Berdasarkan praktikum mengenai struktur kontrol perulangan dan perulangan bersarang (nested loop) yang telah dilaksanakan, dapat ditarik beberapa kesimpulan sebagai berikut:
