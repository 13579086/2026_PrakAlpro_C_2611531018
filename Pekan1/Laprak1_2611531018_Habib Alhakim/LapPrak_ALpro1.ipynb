# BAB II
# PEMBAHASAN

## 2.1 Dasar Teori
1. Bahasa Pemrograman Python
    Python adalah bahasa pemrograman tingkat tinggi yang bersifat interpreted dan berorientasi objek. Python banyak digunakan dalam dunia pendidikan dan industri karena sintaksnya yang bersih, ringkas, dan menyerupai bahasa Inggris manusia. Hal ini membuat Python sangat cocok bagi pemula untuk mempelajari logika dasar pemrograman sebelum beralih ke bahasa yang lebih kompleks.
2. Fungsi Output dan Karakter Khusus
    Fungsi standar untuk menampilkan keluaran atau informasi ke layar dalam Python adalah fungsi print(). Fungsi ini secara otomatis menambahkan baris baru di akhir cetakan teks. Untuk menampilkan karakter khusus seperti backslash (\), kutip, atau membuat tabulasi, Python menyediakan komponen yang disebut escape character. Karakter khusus ini diawali dengan tanda backslash (contoh: \\ untuk mencetak satu backslash tunggal di layar).
3. Komentar dan Manipulasi String
    Komentar adalah baris kode yang diabaikan oleh penerjemah Python saat program dijalankan. Komentar digunakan untuk mendokumentasikan kode agar lebih mudah dipahami oleh pembaca manusia. Python mendukung komentar satu baris menggunakan tanda pagar (#) dan komentar multi-baris menggunakan tiga tanda kutip ("""). Selain itu, Python mengizinkan penggabungan beberapa teks menjadi satu kesatuan melalui proses string concatenation dengan menggunakan operator tambah (+).

## 2.2 Kode Program

### 2.2.1 Program "NamaKelas_NIM"

```python
#program pertama
"""
komentar
dua baris atau lebih
pertemuan 1 praktikum labor
"""
print("Hello World")
print()
print("Departemen Informatika")
print("Universitas Andalas")
print("/ \\ // \\\\ /// \\\\\\")
print("program ini" + " dibuat oleh a\\n " + "Habib Alhakim 2611531018")
```

**Penjelasan :**
1. Bagian pertama kode ini berfungsi untuk mendokumentasikan informasi non-eksekusi melalui komponen komentar. Tanda pagar (#) digunakan untuk mendeklarasikan komentar satu baris yang menandai identitas program. Sementara itu, tanda tiga kutip (""") digunakan untuk menyusun komentar multi-baris (docstring) yang mencakup keterangan praktikum. Seluruh teks di dalam bagian ini diabaikan oleh interpreter Python dan murni berfungsi sebagai dokumentasi kode agar mudah dibaca oleh manusia.
2. Bagian kedua berfungsi untuk menampilkan output informasi standar ke layar menggunakan fungsi cetak print(). Baris-baris ini digunakan untuk memunculkan teks identitas instansi mahasiswa, yaitu "Hello World", "Departemen Informatika", dan "Universitas Andalas". Di antara teks tersebut, terdapat pemanggilan fungsi print() kosong tanpa argumen yang secara spesifik bertujuan untuk memberikan spasi atau jeda satu baris baru (new line) agar tampilan output terlihat lebih rapi dan terstruktur.
3. Bagian ketiga berfungsi untuk melakukan manipulasi string tingkat lanjut melalui penanganan karakter khusus (escape character) dan penggabungan teks (string concatenation). Fungsi print("/ \\ // \\\\ /// \\\\\\") memanfaatkan double backslash (\\) agar karakter backslash tunggal tetap dapat dicetak secara visual ke layar tanpa memicu error sintaks. Selanjutnya, operator tambah (+) digunakan untuk menggabungkan potongan string nama mahasiswa dengan variabel NIM secara horizontal menjadi satu baris output yang utuh.
